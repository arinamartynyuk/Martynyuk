### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30], [20, 30, 40], [30, 40, 50]])
mu = A.mean(axis=0)
sigma = A.std(axis=0)
z = (A - mu) / sigma
print('Column means:', mu)
print('Column std:', sigma)
print('Normalized:\n', z)
````

Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
np.random.seed(42)
M = np.random.randint(1, 10, (4, 4))
row_norms = np.linalg.norm(M, axis=1, keepdims=True)
normalized_rows = M / row_norms
print('Matrix:\n', M)
print('Row norms:', row_norms.ravel())
print('Normalized rows:\n', normalized_rows)
print('Normalized row norms:', np.linalg.norm(normalized_rows, axis=1))
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([[1, 2, 1], [2, 1, -1], [3, -1, 2]])
b = np.array([9, 8, 3])
x = np.linalg.solve(A, b)
print('Solution [x, y, z]:', x)
print('A @ x:', A @ x)
print('b:', b)
print('Equal:', np.allclose(A @ x, b))
````
